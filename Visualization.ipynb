```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("FINAL_USO.csv")

# Convert date column
df['Date'] = pd.to_datetime(df['Date'])

# Set style
sns.set(style="whitegrid")

df.head()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df['Date'], df['Close'])
plt.title("Oil Price Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Oil Close Price")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Close'], bins=30)
plt.title("Distribution of Oil Close Price")
plt.xlabel("Close Price")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(14,10))
corr = df[['Open','High','Low','Close','Volume','USO_Close','GDX_Close']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(df['Close'], df['GDX_Close'])
plt.title("Oil Price vs Gold ETF Price")
plt.xlabel("Oil Close Price")
plt.ylabel("Gold ETF Close Price")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(data=df[['Open','High','Low','Close']])
plt.title("Oil Price Distribution")
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df['Date'], df['Volume'])
plt.title("Oil Trading Volume Over Time")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.show()
```

```python
df['MA50'] = df['Close'].rolling(window=50).mean()
df['MA200'] = df['Close'].rolling(window=200).mean()

plt.figure(figsize=(12,6))
plt.plot(df['Date'], df['Close'], label='Close Price')
plt.plot(df['Date'], df['MA50'], label='50 Day MA')
plt.plot(df['Date'], df['MA200'], label='200 Day MA')

plt.legend()
plt.title("Oil Price with Moving Averages")
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df['Date'], df['High'], label="High")
plt.plot(df['Date'], df['Low'], label="Low")
plt.plot(df['Date'], df['Close'], label="Close")

plt.legend()
plt.title("Oil Price Movement")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x=df['EU_Trend'])
plt.title("EU Trend Distribution")
plt.show()
```
